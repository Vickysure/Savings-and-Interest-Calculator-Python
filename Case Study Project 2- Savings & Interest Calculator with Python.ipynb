```python
# the code below is to create initial customer variables.

customer_name = 'David Okoro'
customer_ID = 'CUST001'
initial_savings = 500000
monthly_contribution = 50000
interest_rate = 5
savings_period_in_months = savings_period = period = 12
savings_goal = 1500000
savings_plan = 'Standard Plan'

# the code below is to check if the initial savings is valid or not.
# initial savings is valid if the amount is greater than or equal to zero.
valid_initial_savings = initial_savings >= 0
print ('Initial Savings is Valid:',valid_initial_savings)
```

```python
# the code below is to check if the monthly contribution is valid.
# a monthly contribution is valid if the amount is greater or equal to Zero.
valid_monthly_contribution = monthly_contribution >= 0
print ('Monthly Contribution is Valid:',valid_monthly_contribution)
```

```python
# the code below is to check if the savings period is valid.
# a savings period is valid if the period is greater than zero and less than 60 months.
valid_savings_period = 0 <= period <= 60
print ('Savings Period is Valid:',valid_savings_period)
```

```python
# the code below is to determine the interest rate of the initial customer.
# the interest rate is determined based on the savings plan:
# - Basic Plan  = 3%
# - Standard Plan = 5%
# - Premium Plan = 7%
# anything else outside these plans is considered an invalid savings plan.

if interest_rate == 3:
    print ('Customer Operates Basic Savings Plan')

elif interest_rate == 5:
    print ('Customer Operates Standard Savings Plan')

elif interest_rate == 7:
    print ('Customer Operates Premium Savings Plan')

else:
    print ('Invalid Savings Plan')
```

```python
# the code below is to calculate the annual interest of the customer.
# Annual_interest is calculated by : (savings_balance * interest_rate) / 100

annual_interest = (initial_savings * interest_rate)/ 100
print (f'Calculated Annual Interest: #{annual_interest}')
```

```python
# the code below is to calculate the monthly interest of the customer.
# monthly interest is calculated by: Annual_interest / 12

monthly_interest = annual_interest / 12
print (f'Calculated Monthly Interest: #{monthly_interest:.2f}')
```

```python
# the code below is to calculate the savings growth.
# the savings growth is calculated by : (New Balance = Previous Balance + Monthly Contribution + Monthly Interest) * period
# the variable 'new_balance' is introduced as the customer's balance after each month of savings
# the varaiable 'final_balance is also introduced as the accumulated savings balance after the savings period of the customer expires.

attempts = 1
while attempts < 13:
    month = int(input ('After Month:'))
    new_balance = initial_savings + monthly_contribution + monthly_interest
    print (f'Balance: #{new_balance:.2f}')
    print ('-' * 45)
    initial_savings = new_balance
    attempts += 1

final_balance = new_balance
print (f'Final Balance after Savings Period: #{final_balance:.2f}')
```

```python
# the code below is to determine the customer's savings goal status.
# customer savings goal is achieved if the final balance amount is greater or equal to customer's savings goal amount.

if final_balance >= savings_goal:
    print ('Customer\'s Savings Goal Achieved.')

else:
    print ('Customer\'s Savings Goal Not Achieved.')
           
```

```python
# the code below is to create a function for the interest amount on the savings of the customer.
# the function will receive final_balance and interest_rate.

def calculate_interest ():
    """the interest is calculated by multiplying the final_balance with the interest_rate, and then dividing the result by 100"""
    interest = (final_balance * interest_rate) /100
    return interest


calculate_interest ()
```
